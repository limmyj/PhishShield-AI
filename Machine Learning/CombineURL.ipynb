```python
def normalize_url(url):
    try:
        # Check if the url is a string before processing
        if isinstance(url, str):
            parsed = urlparse(url.strip().lower())
            netloc = parsed.netloc.lstrip("www.")
            path = parsed.path.rstrip("/")
            normalized = urlunparse((parsed.scheme, netloc, path, '', '', ''))
            return normalized
        else:
            # Return the original value if it's not a string
            return url
    except Exception as e:
        # Return the original value if any other error occurs during parsing
        print(f"Error normalizing URL: {url} - {e}") # Optional: print error for debugging
        return url

# Apply normalization
combined_final['normalized_url'] = combined_final['url'].apply(normalize_url)

# Check how many URLs are duplicate after normalization
duplicates_count = combined_final.duplicated(subset='normalized_url').sum()
print(f"🔍 Duplicate URLs found after normalization: {duplicates_count}")
```

```python
duplicate_rows = combined_final[combined_final.duplicated(keep=False)]
print(f"🔁 Total row-wise duplicates: {len(duplicate_rows)}")
display(duplicate_rows.sort_values(by='url').head(20))
```

```python
# Drop duplicates
deduped_df = combined_final.drop_duplicates(subset='normalized_url', keep='first')

# Drop helper column
deduped_df.drop(columns='normalized_url', inplace=True)

# Save cleaned dataset
deduped_df.to_csv('phishURL1000.csv', index=False)
print(f"✅ Cleaned dataset saved. Total rows: {len(deduped_df)}")
```

```python
print("🔍 Final label distribution after deduplication:")
print(deduped_df['label'].value_counts())

sns.countplot(x='label', data=deduped_df)
plt.xlabel('Label')
plt.ylabel('Count')
plt.title('Label Distribution After Deduplication')
plt.show()
```
